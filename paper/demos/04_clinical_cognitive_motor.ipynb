# Demo 4: Clinical-Cognitive-Motor Phenotyping

**Research Question**: Can we predict cognitive decline from motor symptoms and vice versa?

**Multi-Modal Integration**: UPDRS + MoCA + Cognitive tests over time

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import sys

sys.path.insert(0, str(Path('../..').resolve()))
from src.data_loader import load_ppmi_data, filter_by_cohort
from src.data_preprocessing import extract_feature_groups

sns.set_style('whitegrid')
```

```python
df = load_ppmi_data()
df_pd = filter_by_cohort(df, 'PD')

# Extract features
updrs_cols = extract_feature_groups(df_pd, 'updrs')
cognitive_cols = extract_feature_groups(df_pd, 'cognitive')

print(f"Motor (UPDRS): {len(updrs_cols)} features")
print(f"Cognitive: {len(cognitive_cols)} features")
```

```python
# Longitudinal trajectories
if 'YEAR' in df_pd.columns and updrs_cols and cognitive_cols:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Motor progression
    motor_traj = df_pd.groupby('YEAR')[updrs_cols[0]].mean()
    axes[0].plot(motor_traj.index, motor_traj.values, marker='o', linewidth=2)
    axes[0].set_title('Motor Decline (UPDRS)')
    axes[0].set_xlabel('Years')
    axes[0].set_ylabel(updrs_cols[0])
    
    # Cognitive progression
    cog_traj = df_pd.groupby('YEAR')[cognitive_cols[0]].mean()
    axes[1].plot(cog_traj.index, cog_traj.values, marker='o', linewidth=2, color='coral')
    axes[1].set_title('Cognitive Decline')
    axes[1].set_xlabel('Years')
    axes[1].set_ylabel(cognitive_cols[0])
    
    plt.tight_layout()
    plt.show()
```
