# Demo 5: Prodromal Detection via Multi-Modal Fusion

**Research Question**: Can prodromal patients be distinguished from controls using baseline multi-modal data?

**Multi-Modal Integration**: Clinical + DATSCAN + Biomarkers + Olfaction (UPSIT) at baseline

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys

sys.path.insert(0, str(Path('../..').resolve()))
from src.data_loader import load_ppmi_data, filter_by_cohort, filter_by_visit
```

```
df = load_ppmi_data()
df_baseline = filter_by_visit(df, 'BL')
df_prodromal = filter_by_cohort(df_baseline, ['Prodromal', 'HC'])

print(f"Baseline samples: {len(df_baseline)}")
print(f"\nCohort distribution:")
print(df_prodromal['COHORT'].value_counts())
```

```
# Multi-modal features for early detection
from src.data_preprocessing import extract_feature_groups

features = {}
for group in ['imaging', 'biomarker', 'updrs', 'cognitive']:
    features[group] = extract_feature_groups(df_prodromal, group)
    print(f"{group}: {len(features[group])} features")
```

```
# UPSIT (smell test) analysis
upsit_cols = [col for col in df_prodromal.columns if 'upsit' in col.lower()]

if upsit_cols:
    upsit_col = upsit_cols[0]
    df_prodromal.boxplot(column=upsit_col, by='COHORT', figsize=(8, 6))
    plt.title('Olfactory Function (UPSIT) by Cohort')
    plt.suptitle('')
    plt.ylabel(upsit_col)
    plt.show()
```
